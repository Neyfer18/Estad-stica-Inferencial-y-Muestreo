```python
# %% Cell 1 - Imports y ajustes generales (ejecutar una vez)
%matplotlib inline
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import shapiro, t as t_dist, f as f_dist, probplot
from scipy import stats
from scipy import linalg
import warnings
warnings.filterwarnings('ignore')
sns.set(style="whitegrid")
plt.rcParams['figure.dpi'] = 100

# Helper: safe shapiro (Shapiro requires n>=3)
def safe_shapiro(arr):
    arr = np.asarray(arr).astype(float)
    if len(arr) < 3:
        return None  # not enough data
    try:
        return shapiro(arr)
    except Exception:
        return None

# %% Cell 2 - Función principal: análisis completo (muy detallado)
def analyze_regression(df, x_col, y_col, alfa=0.05, title=None, verbose=True):
    """
    Ejecuta TODO el protocolo de análisis de regresión lineal simple.
    df: pandas DataFrame con las columnas x_col y y_col
    x_col, y_col: nombres de columnas
    alfa: nivel de significancia
    title: título del análisis (opcional)
    """
    # Preparación
    x = df[x_col].astype(float)
    y = df[y_col].astype(float)
    n = len(df)
    if title is None:
        title = f"Análisis: {x_col} vs {y_col}"
    header_sep = "=" * 70

    # ENCABEZADO
    if verbose:
        print(header_sep)
        print(f"ANÁLISIS DE REGRESIÓN LINEAL SIMPLE — {title}")
        print(header_sep)
        print(f"Datos cargados: {n} observaciones")
        print(f"Variable independiente (X): {x_col}")
        print(f"Variable dependiente (Y): {y_col}")
        print(f"Nivel de significancia (α): {alfa}")
        print(header_sep)

    # ESTADÍSTICOS DESCRIPTIVOS
    mean_x = x.mean(); mean_y = y.mean()
    std_x = x.std(ddof=1); std_y = y.std(ddof=1)
    min_x, max_x = x.min(), x.max()
    min_y, max_y = y.min(), y.max()

    if verbose:
        print("\n--- Estadísticos Descriptivos ---")
        print(f"x: media={mean_x:.4f}, sd={std_x:.4f}, min={min_x}, max={max_x}")
        print(f"y: media={mean_y:.4f}, sd={std_y:.4f}, min={min_y}, max={max_y}")

    # GRAFICOS: Histogramas y scatter
    fig = plt.figure(figsize=(14,6))
    ax1 = fig.add_subplot(1,2,1)
    sns.histplot(x, kde=True, ax=ax1)
    ax1.set_title(f'Histograma: {x_col}')
    ax2 = fig.add_subplot(1,2,2)
    sns.histplot(y, kde=True, ax=ax2)
    ax2.set_title(f'Histograma: {y_col}')
    plt.tight_layout()
    plt.show()
    plt.close(fig)

    # Scatter + regression line (simple visual)
    fig, ax = plt.subplots(figsize=(8,6))
    sns.scatterplot(x=x, y=y, s=80, alpha=0.7, ax=ax)
    ax.set_title(f'Diagrama de Dispersión: {x_col} vs {y_col}')
    ax.set_xlabel(x_col); ax.set_ylabel(y_col)
    # compute regression via least squares (robust numerically)
    Xmat = np.column_stack([np.ones(n), x])
    beta_hat = linalg.lstsq(Xmat, y)[0]  # [b0, b1]
    x_line = np.linspace(x.min(), x.max(), 100)
    y_line = beta_hat[0] + beta_hat[1] * x_line
    ax.plot(x_line, y_line, color='red', linewidth=2, label=f'Ŷ = {beta_hat[0]:.3f} + {beta_hat[1]:.3f}X')
    ax.legend()
    plt.grid(True, alpha=0.3)
    plt.show()
    plt.close(fig)

    # PRUEBAS DE NORMALIDAD Shapiro-Wilk (si posible)
    sh_x = safe_shapiro(x)
    sh_y = safe_shapiro(y)

    if verbose:
        print("\n--- Pruebas de Normalidad (Shapiro-Wilk) ---")
        if sh_x is None:
            print(f"x ({x_col}): no hay suficientes observaciones para Shapiro (n={n}) o fallo.")
        else:
            print(f"x ({x_col}): W={sh_x.statistic:.4f}, p={sh_x.pvalue:.4f}")
        if sh_y is None:
            print(f"y ({y_col}): no hay suficientes observaciones para Shapiro (n={n}) o fallo.")
        else:
            print(f"y ({y_col}): W={sh_y.statistic:.4f}, p={sh_y.pvalue:.4f}")

    # CORRELACIONES
    pearson_corr = x.corr(y, method='pearson')
    spearman_corr = x.corr(y, method='spearman')
    kendall_corr = x.corr(y, method='kendall')

    if verbose:
        print("\n--- Coeficientes de Correlación ---")
        print(f"Pearson: {pearson_corr:.4f}")
        print(f"Spearman: {spearman_corr:.4f}")
        print(f"Kendall: {kendall_corr:.4f}")
        # Recomendación
        if (sh_x is not None and sh_x.pvalue > alfa) and (sh_y is not None and sh_y.pvalue > alfa):
            print("Ambas variables aparentan normalidad → Pearson es apropiado.")
        else:
            print("Al menos una variable no normal → usar Spearman/Kendall (no paramétricos).")

    # PRUEBA DE HIPÓTESIS PARA CORRELACIÓN (H0: ρ=0 vs H1: ρ!=0)
    # Sólo válida si n>2
    if n > 2 and np.isfinite(pearson_corr):
        r = pearson_corr
        gl = n - 2
        try:
            t_stat = r * np.sqrt(gl) / np.sqrt(1 - r**2)
            p_val = t_dist.sf(abs(t_stat), gl) * 2
            t_crit = t_dist.ppf(1 - alfa/2, gl)
        except Exception:
            t_stat = np.nan
            p_val = np.nan
            t_crit = np.nan
        if verbose:
            print("\n" + "-"*60)
            print("Prueba de hipótesis para el coeficiente de correlación (two-tailed)")
            print(f"t calculado = {t_stat:.4f}, gl = {gl}, t_crit = ±{t_crit:.4f}, p = {p_val:.4f}")
            print("Decisión:", "Rechazar H0 (ρ≠0)" if p_val < alfa else "No rechazar H0 (no evidencia de correlación)")
    else:
        if verbose:
            print("\nNo hay suficientes observaciones para la prueba de hipótesis de correlación (n<=2).")

    # REGRESIÓN: usando formulación clásica con matriz X
    # b = (X'X)^{-1} X'y
    XtX = Xmat.T.dot(Xmat)
    try:
        inv_XtX = linalg.inv(XtX)
    except Exception:
        inv_XtX = np.linalg.pinv(XtX)
    b = inv_XtX.dot(Xmat.T).dot(y)
    b0, b1 = b[0], b[1]

    # Predicciones, residuos y s_yx
    y_hat = Xmat.dot(b)
    residuos = y - y_hat
    SSE = np.sum(residuos**2)
    SS_Total = np.sum((y - mean_y)**2)
    SSR = SS_Total - SSE
    s_yx = np.sqrt(SSE / (n - 2)) if n > 2 else np.nan

    if verbose:
        print("\n" + "-"*60)
        print("--- RECTA DE MÍNIMOS CUADRADOS ---")
        print(f"b0 (intersección) = {b0:.6f}")
        print(f"b1 (pendiente)    = {b1:.6f}")
        print(f"ECUACIÓN: Ŷ = {b0:.6f} + {b1:.6f} X")
        print(f"s_y,x (Error estándar de estimación) = {s_yx:.6f}")

    # PRUEBA DE SIGNIFICANCIA DE LA PENDIENTE
    if n > 2:
        suma_desviaciones_x = np.sum((x - mean_x)**2)
        s_b = s_yx / np.sqrt(suma_desviaciones_x) if suma_desviaciones_x > 0 else np.nan
        t_pend = (b1 - 0) / s_b if s_b and s_b>0 else np.nan
        gl_pend = n - 2
        t_crit_pend = t_dist.ppf(1 - alfa, gl_pend)  # cola derecha by default
        p_val_pend = t_dist.sf(t_pend, gl_pend) if np.isfinite(t_pend) else np.nan
        if verbose:
            print("\n--- Prueba de significancia de la pendiente ---")
            print(f"t calculado = {t_pend:.4f}, gl = {gl_pend}")
            print(f"Valor p (una cola, H1: b1 > 0) = {p_val_pend:.6f}")
            print("Decisión:", "Rechazar H0 (pendiente significativa)" if p_val_pend < alfa else "No rechazar H0")
    else:
        s_b = np.nan; t_pend = np.nan; p_val_pend = np.nan; t_crit_pend = np.nan

    # COEFICIENTE DE DETERMINACIÓN R^2
    r_squared = 1 - (SSE / SS_Total) if SS_Total != 0 else np.nan
    if verbose:
        print("\n--- R² y Sumas de Cuadrados ---")
        print(f"SS Total = {SS_Total:.4f}, SSR = {SSR:.4f}, SSE = {SSE:.4f}")
        print(f"R² = {r_squared:.4f} (Verificación r^2 ~ pearson^2 = {pearson_corr**2:.4f})")

    # TABLA ANOVA
    if n > 2:
        MSR = SSR / 1
        MSE = SSE / (n - 2)
        F_stat = MSR / MSE if MSE != 0 else np.nan
        p_val_F = f_dist.sf(F_stat, 1, n - 2) if np.isfinite(F_stat) else np.nan
        if verbose:
            print("\n--- Tabla ANOVA ---")
            print(f"{'Fuente':<12}{'SS':>12}{'gl':>8}{'MS':>12}{'F':>12}{'p-val':>12}")
            print(f"{'Regresión':<12}{SSR:12.4f}{1:8d}{MSR:12.4f}{F_stat:12.4f}{p_val_F:12.6f}")
            print(f"{'Error':<12}{SSE:12.4f}{n-2:8d}{MSE:12.4f}")
            print(f"{'Total':<12}{SS_Total:12.4f}{n-1:8d}")
            print("Modelo significativo?" , "Sí" if p_val_F < alfa else "No")
    else:
        F_stat = np.nan; p_val_F = np.nan

    # INTERVALOS DE CONFIANZA Y PREDICCIÓN (para una lista de valores)
    t_crit_interval = t_dist.ppf(1 - alfa/2, n-2) if n>2 else np.nan
    valores_x_ejemplo = np.linspace(x.min(), x.max(), 5) if n>0 else []

    ic_list = []
    ip_list = []
    for x_val in valores_x_ejemplo:
        y_pred = b0 + b1 * x_val
        termino_conf = np.sqrt((1/n) + ((x_val - mean_x)**2 / suma_desviaciones_x)) if suma_desviaciones_x>0 else np.nan
        margen_conf = t_crit_interval * s_yx * termino_conf if np.isfinite(t_crit_interval) else np.nan
        ic_list.append((x_val, y_pred, y_pred - margen_conf, y_pred + margen_conf))

        termino_pred = np.sqrt(1 + (1/n) + ((x_val - mean_x)**2 / suma_desviaciones_x)) if suma_desviaciones_x>0 else np.nan
        margen_pred = t_crit_interval * s_yx * termino_pred if np.isfinite(t_crit_interval) else np.nan
        ip_list.append((x_val, y_pred, y_pred - margen_pred, y_pred + margen_pred))

    if verbose:
        print("\n--- Intervalos de Confianza (para la media) y de Predicción (individuales) ---")
        print(f"t crítico para intervalos (dos colas): ±{t_crit_interval:.4f}")
        print("\nEjemplos (X, ŷ, IC_inf, IC_sup)")
        for row in ic_list:
            print(f"{row[0]:.3f} -> ŷ={row[1]:.3f} , IC=[{row[2]:.3f}, {row[3]:.3f}]")
        print("\nEjemplos IP (individuales):")
        for row in ip_list:
            print(f"{row[0]:.3f} -> ŷ={row[1]:.3f} , IP=[{row[2]:.3f}, {row[3]:.3f}]")

    # GRAFICO: recta con IC e IP sobre rango
    if n > 2:
        x_range = np.linspace(x.min() - 0.05*(x.max()-x.min()+1), x.max() + 0.05*(x.max()-x.min()+1), 200)
        ic_lower = []; ic_upper = []; ip_lower = []; ip_upper = []
        for xv in x_range:
            ypred = b0 + b1 * xv
            term_conf = np.sqrt((1/n) + ((xv - mean_x)**2 / suma_desviaciones_x))
            me_conf = t_crit_interval * s_yx * term_conf
            ic_lower.append(ypred - me_conf); ic_upper.append(ypred + me_conf)
            term_pred = np.sqrt(1 + (1/n) + ((xv - mean_x)**2 / suma_desviaciones_x))
            me_pred = t_crit_interval * s_yx * term_pred
            ip_lower.append(ypred - me_pred); ip_upper.append(ypred + me_pred)

        fig, ax = plt.subplots(figsize=(10,6))
        ax.scatter(x, y, s=80, alpha=0.7, label='Datos')
        ax.plot(x_range, b0 + b1 * x_range, 'r-', label='Recta de Regresión')
        ax.fill_between(x_range, ic_lower, ic_upper, alpha=0.25, label=f'IC {(1-alfa)*100:.0f}% para la Media')
        ax.fill_between(x_range, ip_lower, ip_upper, alpha=0.15, label=f'IP {(1-alfa)*100:.0f}% Individual')
        ax.set_title('Regresión con IC e IP')
        ax.set_xlabel(x_col); ax.set_ylabel(y_col)
        ax.legend(loc='upper left')
        plt.grid(True, alpha=0.3)
        plt.show()
        plt.close(fig)

    # ANALISIS DE RESIDUOS
    df_res = df.copy()
    df_res = df_res.assign(y_predicho=y_hat, residuos=residuos)
    df_res['res_est'] = df_res['residuos'] / s_yx if np.isfinite(s_yx) else np.nan

    if verbose:
        print("\n--- Análisis de Residuos ---")
        print(df_res[[x_col, y_col, 'y_predicho', 'residuos', 'res_est']].to_string(index=False))

    # Graficos de residuos: 4 plots
    fig, axes = plt.subplots(2,2, figsize=(14,10))
    axes = axes.flatten()
    axes[0].scatter(df_res['y_predicho'], df_res['residuos'], s=80, alpha=0.7)
    axes[0].axhline(0, color='red', linestyle='--')
    axes[0].set_title('Residuos vs Valores Predichos')
    axes[0].set_xlabel('ŷ'); axes[0].set_ylabel('Residuos')

    axes[1].scatter(df_res[x_col], df_res['residuos'], s=80, alpha=0.7)
    axes[1].axhline(0, color='red', linestyle='--')
    axes[1].set_title('Residuos vs X')
    axes[1].set_xlabel(x_col); axes[1].set_ylabel('Residuos')

    axes[2].hist(df_res['residuos'], bins=max(4, int(n/2)), alpha=0.7, edgecolor='black')
    axes[2].axvline(0, color='red', linestyle='--')
    axes[2].set_title('Histograma de Residuos')

    probplot(df_res['residuos'], dist="norm", plot=axes[3])
    axes[3].set_title('Q-Q Plot de Residuos')
    plt.tight_layout()
    plt.show()
    plt.close(fig)

    # Shapiro para residuos (si posible)
    sh_res = safe_shapiro(df_res['residuos'])
    if verbose:
        if sh_res is None:
            print("\nNo hay suficientes observaciones para Shapiro sobre residuos o fallo.")
        else:
            print(f"\nShapiro residuos: W={sh_res.statistic:.4f}, p={sh_res.pvalue:.4f}")
            print("Normalidad residuos:", "Se cumple" if sh_res.pvalue > alfa else "No se cumple")

    # Outliers
    outliers = df_res[np.abs(df_res['res_est']) > 2.5]
    if verbose:
        print(f"\nOutliers (|res_est| > 2.5): {len(outliers)}")
        if len(outliers) > 0:
            print(outliers[[x_col, y_col, 'res_est']])

    # EVALUACIÓN TRANSFORMACIÓN
    necesita_transform = False
    razones = []
    if (sh_x is not None and sh_x.pvalue <= alfa) or (sh_y is not None and sh_y.pvalue <= alfa):
        razones.append("- Variables originales no normales")
        necesita_transform = True
    if (sh_res is not None and sh_res.pvalue <= alfa):
        razones.append("- Residuos no normales")
        necesita_transform = True
    if np.abs(pearson_corr) < 0.5:
        razones.append("- Correlación lineal débil (|r| < 0.5)")
        necesita_transform = True
    if len(outliers) > n * 0.15:
        razones.append(f"- Exceso de outliers ({len(outliers)} casos > 15%)")
        necesita_transform = True

    if verbose:
        print("\n--- Evaluación de necesidad de transformación ---")
        if necesita_transform:
            print("Se recomienda considerar transformaciones por las siguientes razones:")
            for r_ in razones:
                print("  ", r_)
            print("Transformaciones sugeridas: log, sqrt, reciprocal, Box-Cox (si >0).")
        else:
            print("No se recomienda transformación. Supuestos razonables.")

    # RESUMEN EJECUTIVO (conciso)
    if verbose:
        print("\n" + "-"*60)
        print("RESUMEN EJECUTIVO (resumen):")
        print(f"Ecuación: Ŷ = {b0:.4f} + {b1:.4f} X")
        print(f"r (Pearson) = {pearson_corr:.4f} ; R² = {r_squared:.4f}")
        print(f"Prueba pendiente: t = {t_pend if np.isfinite(t_pend) else 'NA'}, p = {p_val_pend if np.isfinite(p_val_pend) else 'NA'}")
        print(f"ANOVA: F = {F_stat if np.isfinite(F_stat) else 'NA'}, p = {p_val_F if np.isfinite(p_val_F) else 'NA'}")
        print("Conclusión (indicativa):", end=" ")
        if np.isfinite(p_val_pend) and p_val_pend < alfa and np.isfinite(p_val_F) and p_val_F < alfa:
            print("El modelo es significativo y útil para predicción.")
        else:
            print("El modelo puede tener limitaciones; considere más datos o variables.")
        print(header_sep)

    # Devuelvo un diccionario con resultados clave por si el usuario quiere usarlo programáticamente
    results = {
        'n': n,
        'mean_x': mean_x, 'mean_y': mean_y,
        'std_x': std_x, 'std_y': std_y,
        'pearson': pearson_corr, 'spearman': spearman_corr, 'kendall': kendall_corr,
        'b0': b0, 'b1': b1,
        'SSE': SSE, 'SSR': SSR, 'SST': SS_Total,
        's_yx': s_yx,
        'r_squared': r_squared,
        't_corr': t_stat if 't_stat' in locals() else None, 'p_corr': p_val if 'p_val' in locals() else None,
        't_pend': t_pend, 'p_pend': p_val_pend,
        'F_stat': F_stat, 'p_F': p_val_F,
        'residuals_df': df_res,
        'ic_examples': ic_list, 'ip_examples': ip_list,
        'need_transform': necesita_transform, 'reasons_transform': razones
    }
    return results

# %% Cell 3 - Ejercicio 1 (Goles anotados vs recibidos) [desde PDF]
df1 = pd.DataFrame({
    'Equipo': ['Leones','Águilas','Tigres','Toros','Pumas','Jaguares'],
    'Goles_anotados': [38,45,31,50,27,41],
    'Goles_recibidos': [22,28,39,33,36,29]
})
# Ejecutar análisis (llama la función)
res1 = analyze_regression(df1[['Goles_anotados','Goles_recibidos']], 'Goles_anotados','Goles_recibidos',
                           alfa=0.05, title='Ejercicio 1 - Goles (X=Anotados, Y=Recibidos)')

# %% Cell 4 - Ejercicio 2 (Horas estudio vs Calificación) [12 estudiantes]
df2 = pd.DataFrame({
    'Horas_estudio':[4,7,10,6,8,9,5,11,7,6,8,10],
    'Calificacion':[65,70,85,78,82,91,74,94,80,77,84,88]
})
res2 = analyze_regression(df2, 'Horas_estudio','Calificacion', alfa=0.05, title='Ejercicio 2 - Horas vs Calificación')

# %% Cell 5 - Ejercicio 3 (Llamadas diarias vs Ventas semanales)
df3 = pd.DataFrame({
    'Llamadas_diarias':[15,20,18,25,22,17,19,23,21,16],
    'Ventas_semanales':[5,7,6,11,9,6,7,10,8,5]
})
res3 = analyze_regression(df3, 'Llamadas_diarias','Ventas_semanales', alfa=0.05, title='Ejercicio 3 - Llamadas vs Ventas')

# %% Cell 6 - Ejercicio 4 (The Cotton Mill) - Unir regiones en tres DataFrames o analizar cada región por separado
# Region 1
reg1 = pd.DataFrame({
    'Tienda':[3.7,5.0,7.0,6.5,2.2,5.5,2.9],
    'Ventas':[9.18,8.54,6.85,8.98,5.20,7.15,7.88]
})
# Region 2
reg2 = pd.DataFrame({
    'Tienda':[4.2,3.4,2.5,1.5,3.7,4.9,3.2],
    'Ventas':[5.65,3.26,2.25,1.95,3.34,7.23,4.45]
})
# Region 3
reg3 = pd.DataFrame({
    'Tienda':[2.2,1.6,2.3,2.9,3.2,3.7,4.4,4.8,5.0],
    'Ventas':[2.05,1.43,2.44,2.95,3.16,3.95,4.25,5.39,5.64]
})
print("\n--- Región 1 ---")
res4_reg1 = analyze_regression(reg1, 'Tienda','Ventas', alfa=0.05, title='Ejercicio 4 - Región 1')
print("\n--- Región 2 ---")
res4_reg2 = analyze_regression(reg2, 'Tienda','Ventas', alfa=0.05, title='Ejercicio 4 - Región 2')
print("\n--- Región 3 ---")
res4_reg3 = analyze_regression(reg3, 'Tienda','Ventas', alfa=0.05, title='Ejercicio 4 - Región 3')

# %% Cell 7 - Ejercicio 5 (Crecimiento bacteriano: ajuste + log transform)
df5 = pd.DataFrame({
    'Tiempo':[1,2,3,4,5,6,7],
    'Poblacion':[45,112,228,485,900,1720,3526]
})
print("\n--- Modelo original (Tiempo vs Población) ---")
res5_orig = analyze_regression(df5, 'Tiempo','Poblacion', alfa=0.05, title='Ejercicio 5 - Original')
# Transformación logaritmica (usar log natural)
df5_log = df5.copy()
df5_log['logPobl'] = np.log(df5_log['Poblacion'])
print("\n--- Modelo transformado (Tiempo vs log(Población)) ---")
res5_log = analyze_regression(df5_log, 'Tiempo','logPobl', alfa=0.05, title='Ejercicio 5 - Log transform')

# %% Cell 8 - Ejercicio 6 (Ingreso vs Gasto en alimentos) - modelar original y log-log
df6 = pd.DataFrame({
    'Ingreso':[1.2,1.8,2.4,3.5,5.0,7.5,10.0],
    'Gasto':[2.1,2.7,3.0,4.2,5.0,6.5,7.3]
})
print("\n--- Modelo original (Ingreso vs Gasto) ---")
res6_orig = analyze_regression(df6, 'Ingreso','Gasto', alfa=0.05, title='Ejercicio 6 - Original')
# Transformación log-log (usar log natural)
df6_log = df6.copy()
df6_log['logIngreso'] = np.log(df6_log['Ingreso'])
df6_log['logGasto'] = np.log(df6_log['Gasto'])
print("\n--- Modelo log-log (log(Ingreso) vs log(Gasto)) ---")
res6_loglog = analyze_regression(df6_log, 'logIngreso','logGasto', alfa=0.05, title='Ejercicio 6 - Log-Log')

# %% Cell 9 - Ejercicio 7 (Intensidad vs Concentración) - sqrt transform on Y
df7 = pd.DataFrame({
    'Intensidad':[10,30,50,70,90,110,130],
    'Concentracion':[2.2,3.8,6.1,11.7,19.2,24.2,36.9]
})
print("\n--- Modelo original (Intensidad vs Concentración) ---")
res7_orig = analyze_regression(df7, 'Intensidad','Concentracion', alfa=0.05, title='Ejercicio 7 - Original')
# Transformación raíz cuadrada en Y
df7_sqrt = df7.copy()
df7_sqrt['sqrtConc'] = np.sqrt(df7_sqrt['Concentracion'])
print("\n--- Modelo transformado (Intensidad vs sqrt(Concentración)) ---")
res7_sqrt = analyze_regression(df7_sqrt, 'Intensidad','sqrtConc', alfa=0.05, title='Ejercicio 7 - Sqrt Y')

# %% Cell 10 - Ejercicio 8 (Crecimiento bacteriano Cali) - similar a 5
df8 = pd.DataFrame({
    'Tiempo':[1,2,3,4,5,6,7],
    'Poblacion':[42,110,220,495,910,1730,3500]
})
print("\n--- Modelo original (Tiempo vs Población - Cali) ---")
res8_orig = analyze_regression(df8, 'Tiempo','Poblacion', alfa=0.05, title='Ejercicio 8 - Original')
df8_log = df8.copy()
df8_log['logPobl'] = np.log(df8_log['Poblacion'])
print("\n--- Modelo transformado (Tiempo vs log(Población) - Cali) ---")
res8_log = analyze_regression(df8_log, 'Tiempo','logPobl', alfa=0.05, title='Ejercicio 8 - Log transform')

# %% Cell 11 - Ejercicio 9 (Distancia vs Tarifa - vuelos)
df9 = pd.DataFrame({
    'Distancia':[636,275,398,405,286,627,2346,177,2528,248,512,248,248,237,621,853,2181,1531,2724,999],
    'Tarifa':[109,129,141,152,165,259,231,148,224,125,225,124,125,137,191,191,249,229,243,219]
})
print("\n--- Modelo Distancia vs Tarifa ---")
res9 = analyze_regression(df9, 'Distancia','Tarifa', alfa=0.05, title='Ejercicio 9 - Distancia vs Tarifa')

# %% Cell 12 - (Opcional) Guardar resultados clave a disco (ejemplo)
# Puedes guardar los resultados en CSV o pickle si quieres.
# Ejemplo: guardar residuos del ejercicio 9
try:
    df9_res = res9['residuals_df']
    df9_res.to_csv('/content/ejercicio9_residuos.csv', index=False)
    print("Residuos Ejercicio 9 guardados en /content/ejercicio9_residuos.csv")
except Exception as e:
    print("No se pudo guardar archivos:", e)

# Fin del Notebook
print("\nNotebook completado. Ejecuta las celdas por orden. Si quieres dividir el notebook en celdas, corta donde están los comentarios '# %% Cell X'.")
```
